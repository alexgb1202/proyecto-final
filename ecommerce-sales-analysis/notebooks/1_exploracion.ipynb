## 0. Importamos las librerias que vamos a usar

```python
import pandas as pd
import numpy as np
import os
```

## 1. Unificación previa de tablas antes de la exploración

```python
# 1. Antes de explorar los datos, vamos a unir tablas de la misma categoria para no explorar 16 tablas manualmente

# 1.1 Recoger todos los dataframes de la carpeta 0_raw
ruta = "../Data/0_raw/"
archivos = os.listdir(ruta) # Lista con los nombres de nuestros csv en 0_raw

dfs_raw = {} # Diccionario vacio que contendra todos los df en crudo

for archivo in archivos:
    nombre = archivo.replace(".csv", "") # Quitamos .csv para recoger solo el nombre y usarlo como clave del diccionario
    dfs_raw[nombre] = pd.read_csv(ruta + archivo) # Cargamos el archivo CSV real usando su nombre exacto y lo guardamos en el diccionario.

```

```python
# 1.2 Unir los dataframes de la misma categoria

# 1.2.1 Deuda
deuda = pd.concat([
    dfs_raw['deuda2009'],
    dfs_raw['deuda2010'],
    dfs_raw['deuda2011']],
    ignore_index= True
)

deuda # Podemos previsualizar que las fechas del 2009 no se estan viendo porque no existe la columna Fecha en 2009
```

```python
# 1.2.2 Paro
paro = pd.concat([
    dfs_raw['paro2009'],
    dfs_raw['paro2010'],
    dfs_raw['paro2011']],
    ignore_index= True
)

paro
```

```python
# 1.2.3 pib_per_capita
ppcapita = pd.concat([
    dfs_raw['pib_per_capita2009'],
    dfs_raw['pib_per_capita2010'],
    dfs_raw['pib_per_capita2011']],
    ignore_index= True
)

ppcapita
```

```python
# 1.2.4 poblacion
poblacion = pd.concat([
    dfs_raw['poblacion2009'],
    dfs_raw['poblacion2010'],
    dfs_raw['poblacion2011']],
    ignore_index= True
)

poblacion
```

```python
# 1.2.6 pib
pib = pd.concat([
    dfs_raw['pib2009'],
    dfs_raw['pib2010'],
    dfs_raw['pib2011']],
    ignore_index= True
)

pib
```

```python
# 1.2.6 online_retail_II (archivo en bruto descargado de Kaggle)
retail = pd.read_csv('../Data/0_raw/online_retail_II.csv')
retail

# Nota: El dataset de retail tiene los países en inglés, mientras que los datos macroeconómicos están en español.
# Para poder hacer un merge correcto en la fase de limpieza, será necesario traducir los países de los datos macro al inglés.
```

### 1.3 Mandar los datos a la carpeta Interim

```python
deuda.to_csv('../Data/1_interim/deuda.csv', index = False)
paro.to_csv('../Data/1_interim/paro.csv', index = False)
poblacion.to_csv('../Data/1_interim/poblacion.csv', index = False)
pib.to_csv('../Data/1_interim/pib.csv', index = False)
ppcapita.to_csv('../Data/1_interim/ppcapita.csv', index = False)
retail.to_csv('../Data/1_interim/retail.csv', index = False)
```

## 2. Exploración

Tenemos **6 archivos** para explorar:

- **deuda**
- **paro**
- **población**
- **pib**
- **ppcapita**
- **retail**





```python
# 2.1 Deuda

deuda.info()
# - Eliminar columnas: 'Deuda total (M.€).1', 'Deuda total (M.$)',
#   'Deuda total (M.$).1' y 'Deuda Per Cápita.1'.
# - Convertir a tipo numérico: 'Deuda total (M.€)', 'Deuda (%PIB)' y
#   'Deuda Per Cápita'.
# - Convertir 'Fecha' en 'Año' a numerico.

deuda.sample(5)
# - La columna 'Fecha' tiene NaN para 2009 porque la tabla original no
#   incluía el año. Se imputará el valor 2009.
# - Traducir los nombres de los países y eliminar el sufijo '[+]'
#   para que coincidan con el DataFrame 'retail'.

deuda.describe(include='all')
# - Se confirma que algunas columnas tienen tipos de datos incorrectos.

deuda.isna().sum()
# - Los únicos NaN relevantes corresponden a la columna 'Fecha'.
# - El resto pertenecen a columnas que se eliminarán.

deuda.duplicated().sum()
# - Existe una fila duplicada que deberá eliminarse.

deuda[deuda.duplicated(keep=False)]
# - Visualización de la fila duplicada.
```

```python
# 2.2 Paro

paro.info()
# - Eliminar las columnas: 'Tasa de desempleo.1', 'Var.' y 'Var. Año'.
# - Convertir 'Tasa de desempleo' a tipo numérico.
# - Renombrar la columna 'Mes' a 'Año' y convertirla a tipo int.

paro.sample(5)
paro["Mes"].value_counts()
# - Eliminar algunas filas que se han colado en el scrapeo.
# - Extraer el año de la columna 'Mes', renombrarla a 'Año' y convertirla a int.
# - Eliminar el sufijo '[+]' de los nombres de los países.

paro.describe(include="all")
# - Se confirma que es necesario realizar la limpieza para obtener un resumen
#   estadístico correcto.

paro.isna().sum()
# - Los únicos valores nulos pertenecen a columnas que serán eliminadas
#   ('Tasa de desempleo.1' y 'Var. Año').

paro.duplicated().sum()
# - No existen filas duplicadas.

```

```python
# 2.3 Poblacion

poblacion.info()
# - Eliminaremos columnas como: 'Densidad', 'Población.1' y 'Var.'
# - Cambiaremos el tipo de datos a int de la columna 'Población'

poblacion.sample(5)
# - Eliminar el sufijo '[+]' de los nombres de los países.


poblacion.describe(include="all")
# - Se confirma que es necesario realizar la limpieza para obtener un resumen
#   estadístico correcto.

poblacion.isna().sum()
# - La unica columna que tiene nulos va a ser eliminada porque no la necesitamos

poblacion.duplicated().sum()
# - No hay presencia de duplicados
```

```python
# 2.4 PIB

pib.info()
# - Eliminar las columnas: 'PIB anual.1', 'PIB anual.2', 'PIB anual.3',
#   'Var. PIB (%)' y 'Var. PIB (%).1'.
# - Renombrar la columna 'Fecha' a 'Año' y convertirla a tipo int.
# - Convertir 'PIB anual' a tipo numérico, eliminando previamente los
#   caracteres que podrían impedir la conversión

pib.sample(5)
# - Eliminar el sufijo '[+]' de los nombres de los países.

pib.describe(include="all")
# - Se confirma que es necesario realizar la limpieza para obtener un
#   resumen estadístico correcto.

pib["Países"].value_counts()
# - Eliminar las filas que contienen texto y no corresponden a países,
#   por ejemplo: '< PIB 2008' o 'PIB 2010 >'.

pib.isna().sum()
# - Las columnas 'PIB anual.1', 'PIB anual.3', 'Var. PIB (%)' y
#   'Var. PIB (%).1' contienen valores nulos, pero no es necesario
#   tratarlos porque dichas columnas serán eliminadas.

pib.duplicated().sum()
# - No existen filas duplicadas.
```

```python
# 2.5 ppcapita

ppcapita.info()
# - Cambiar nombre de Fecha a Año y su tipo de datos a int
# - Nos quedamos solo con las columnas: 'Países', 'Año', 'PIB Per Capita'
# - Cambiar el tipo de datos de PIB Per Capita a int

ppcapita.sample(5)
# - Quitar el sufijo [+] de los paises.

ppcapita.describe(include="all")
# - Se confirma que es necesario realizar la limpieza para obtener un
#   resumen estadístico correcto.

ppcapita["Países"].value_counts()
# - Eliminar las filas que contienen texto y no corresponden a países,
#   por ejemplo: '< PIB 2008PIB 2010 >'.

ppcapita.isna().sum()
# - Tenemos nulos en varias columnas. Pero son nulos de columnas que vamos a eliminar


ppcapita.duplicated().sum()
# - No tenemos duplicados
```

```python
# 2.6 retail

retail.info()
# - Revisamos la estructura del dataset para detectar tipos incorrectos y códigos mixtos.
# - InvoiceNo y StockCode se mantienen como object porque las letras contienen información útil.
# - InvoiceDate se usará para crear una jerarquía temporal (año, mes, día) para análisis y merge.
# - CustomerID se convertirá a int para permitir segmentación y análisis por cliente.
# - Evaluaremos la creación de columnas derivadas (IsCancelled, Variant) sin perder información.

retail.sample(5)
# - Nos damos cuenta que no solo tenemos ventas, tambien tenemos movimientos internos del almacen
# con descripciones, customer id, cantidades y precios que sesgan nuestro analisis puramente de ventas

retail.describe().T
# - Quantity presenta valores negativos → devoluciones.
# - Quantity y Price tienen valores extremadamente altos (hasta 80k y 38k) → outliers que requieren revisión.
# - Price negativo → ajustes contables o devoluciones.
# - La presencia de outliers confirma que será necesaria una limpieza cuidadosa antes del análisis final.

retail.isna().sum()
# - Tenemos 243007 valores nulos en customer id que deberemos tratar o imputar
# - Tenemos 4382 valores nulos en description, deberemos imputar con "Sin descripcion"

retail["Country"].value_counts()
# - Necesario para obtener la traduccion de los paises para la posterior traduccion y el merge

retail.duplicated().sum()
# - Tenemos 34335 duplicados, que eliminaremos en la limpieza
```
