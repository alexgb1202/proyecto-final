## 0. Importamos las librerias que vamos a usar

```python
import pandas as pd
import numpy as np
```

## 1. Scrapeamos los datos


```python

# 1.1 Población
años = [2009, 2010, 2011]
poblacion = {}

for año in años:
    df = pd.read_html(f"https://datosmacro.expansion.com/demografia/poblacion?anio={año}")[0]
    df["Año"] = año
    poblacion[año] = df

poblacion[2009]

# Enviar a carpeta Data/raw
for año, df in poblacion.items(): # .items() recorre los pares clave-valor de un diccionario; en este caso: {año: df}
    df.to_csv(f"../Data/0_raw/poblacion{año}.csv", index=False)
```

```python
# 1.2 PIB
años = [2009, 2010, 2011]
pib = {}

for año in años:
    pib[año] = pd.read_html(f'https://datosmacro.expansion.com/pib?anio={año}')[0]

pib[2009]

# Enviar a carpeta Data/raw
for año, df in pib.items():
    pib[año].to_csv(f'../Data/0_raw/pib{año}.csv', index = False)
```

```python
# 1.3 PIB per capita
años = [2009, 2010, 2011]
pib_per_capita = {}
for año in años:
    pib_per_capita[año] = pd.read_html(f'https://datosmacro.expansion.com/pib?anio={año}')[1]

pib_per_capita[2009]

# Enviar a carpeta Data/raw

for año, df in pib_per_capita.items():
    pib_per_capita[año].to_csv(f'../Data/0_raw/pib_per_capita{año}.csv', index = False)
```
